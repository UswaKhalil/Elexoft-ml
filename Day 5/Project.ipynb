```python
class Book:
    def __init__(self, title, author, pages, genre):
        self.title = title
        self.author = author
        self.pages = pages
        self.genre = genre
        self.status = "Not Started"
        self.pages_read = 0

    def read_pages(self, pages):
        self.pages_read += pages

        if self.pages_read >= self.pages:
            self.pages_read = self.pages
            self.status = "Finished"
        else:
            self.status = "Currently Reading"

    def to_dict(self):
        return {
            "title": self.title,
            "author": self.author,
            "pages": self.pages,
            "genre": self.genre,
            "status": self.status,
            "pages_read": self.pages_read
        
        }
```

```python
book1 = Book(
    "The Hobbit",
    "J.R.R. Tolkien",
    310,
    "Fantasy"
)

print("Title:", book1.title)
print("Author:", book1.author)
print("Pages:", book1.pages)
print("Genre:", book1.genre)
print("Status:", book1.status)
print("Pages Read:", book1.pages_read)
```

```python
book1.read_pages(310)

print("Pages Read:", book1.pages_read)
print("Status:", book1.status)
```

```python
class AudioBook(Book):
    def __init__(
        self,
        title,
        author,
        pages,
        genre,
        narrator,
        listening_time
    ):
        super().__init__(title, author, pages, genre)

        self.narrator = narrator
        self.listening_time = listening_time
```

```python
class Library:
    def __init__(self):
        self.books = []
        self.genres = set()

    def add_book(self):
        print("\n--- Add New Book ---")

        title = input("Enter book title: ")
        author = input("Enter author: ")
        genre = input("Enter genre: ")

        while True:
            try:
                pages = int(input("Enter total number of pages: "))

                if pages <= 0:
                    print("Pages must be greater than 0.")
                    continue

                break

            except ValueError:
                print("Please enter a valid number for pages.")

        book = Book(title, author, pages, genre)

        self.books.append(book.to_dict())
        self.genres.add(genre)

        print("Book added successfully!")

    def display_books(self):
        if not self.books:
            print("\nYour library is empty.")
            return

        print("\n========== YOUR LIBRARY ==========")

        for index, book in enumerate(self.books, start=1):
            print(f"\nBook #{index}")
            print(f"Title: {book['title']}")
            print(f"Author: {book['author']}")
            print(f"Genre: {book['genre']}")
            print(f"Pages: {book['pages']}")
            print(f"Pages Read: {book['pages_read']}")
            print(f"Status: {book['status']}")

        print("\n==================================")

    def search_by_author(self):
        author = input("\nEnter author name: ").strip().lower()

        found = False

        for book in self.books:
            if author in book["author"].lower():
                print("\nBook Found:")
                print(f"Title: {book['title']}")
                print(f"Author: {book['author']}")
                print(f"Genre: {book['genre']}")
                print(f"Status: {book['status']}")

                found = True

        if not found:
            print("No books found by that author.")

    def search_by_genre(self):
        genre = input("\nEnter genre: ").strip().lower()

        found = False

        for book in self.books:
            if genre in book["genre"].lower():
                print("\nBook Found:")
                print(f"Title: {book['title']}")
                print(f"Author: {book['author']}")
                print(f"Genre: {book['genre']}")
                print(f"Status: {book['status']}")

                found = True

        if not found:
            print("No books found in that genre.")

    def show_genres(self):
        print("\n========== GENRES ==========")

        if self.genres:
            for genre in sorted(self.genres):
                print("-", genre)
        else:
            print("No genres available.")

        print("============================")

    def save_library(self):
        try:
            with open("library.json", "w") as file:
                json.dump(self.books, file, indent=4)

            print("Library saved successfully!")

        except Exception as e:
            print("Could not save library:", e)

    def load_library(self):
        try:
            with open("library.json", "r") as file:
                self.books = json.load(file)

            self.genres = set()

            for book in self.books:
                self.genres.add(book["genre"])

            print("Library loaded successfully!")

        except FileNotFoundError:
            print("No saved library found. Starting with an empty library.")

        except json.JSONDecodeError:
            print("The library file contains invalid data.")
```

```python
def main_menu():
    library = Library()

    while True:
        print("\n" + "=" * 45)
        print("       PERSONAL LIBRARY TRACKER")
        print("=" * 45)

        print("1. Add a new book")
        print("2. Search by author")
        print("3. Search by genre")
        print("4. Display all books")
        print("5. Show all genres")
        print("6. Load library")
        print("7. Save library")
        print("0. Exit")

        print("=" * 45)

        choice = input("Enter your choice: ")

        if choice == "1":
            library.add_book()

        elif choice == "2":
            library.search_by_author()

        elif choice == "3":
            library.search_by_genre()

        elif choice == "4":
            library.display_books()

        elif choice == "5":
            library.show_genres()

        elif choice == "6":
            library.load_library()

        elif choice == "7":
            library.save_library()

        elif choice == "0":
            print("\nThank you for using Personal Library Tracker!")
            break

        else:
            print("\nInvalid choice. Please enter a number from 0 to 7.")
```
