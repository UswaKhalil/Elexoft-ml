# UFO Sightings — Cleaning, Classing, Encoding & Summary
This notebook takes the raw `Location`, `UFO_shape`, `length_of_encounter_seconds` data and:
1. Fills in missing values
2. Splits `Location` into country / state / city
3. Groups (classifies) cities so rare one-off cities don't clutter the data
4. Converts text categories into 0/1 columns (one-hot encoding)
5. Produces a final summary: which location tends to have which UFO shape, and for how long

## Step 0: Load the data
Make sure `ufo_final.csv` is in the same folder as this notebook (or update the path below).

```python
import pandas as pd

df = pd.read_csv('ufo_final.csv')
print(df.shape)
df.head()
```

## Step 1: Handle missing values
Some rows have no `Location`. Instead of deleting them, we fill them in as `"Unknown"` so we keep as much data as possible while being honest that the location wasn't recorded.

```python
df['Location'] = df['Location'].fillna('Unknown | Unknown | Unknown')

print('Missing Location values now:', df['Location'].isna().sum())
```

## Step 2: Split `Location` into country, state, and city
The `Location` column is one long string like `"us | wa | spokane"`. We split it into 3 separate columns so we can work with each piece individually.

```python
loc_parts = df['Location'].str.split(' | ', expand=True, regex=False)

df['country'] = loc_parts[0].str.strip()
df['state']   = loc_parts[1].str.strip()
df['city']    = loc_parts[2].str.strip()

df[['Location', 'country', 'state', 'city']].head()
```

## Step 3: Make classes for cities

**Why?** There are hundreds of different cities in this dataset, and most of them only appear **once**. If we tried to use every single city as its own category, we'd end up with hundreds of near-empty categories — too messy for any pattern to show up.

**Fix:** Keep the most frequent cities as their own class, and group everything else into `"Other"`. This turns hundreds of cities into a small, manageable set of classes.

```python
city_counts = df['city'].value_counts()
print('Total unique cities:', df['city'].nunique())
print('Cities that appear only once:', (city_counts == 1).sum())

# Keep the top 15 most common cities, group the rest into 'Other'
top_cities = city_counts.nlargest(15).index
df['city_class'] = df['city'].apply(lambda x: x if x in top_cities else 'Other')

df['city_class'].value_counts()
```

## Step 4: Convert category columns into 0s and 1s (one-hot encoding)

Computers can't do math on words like `"triangle"` or `"chicago"`. So for each category, we create a new column that is either `1` (yes, this row is that category) or `0` (no, it isn't).

Example: instead of one `UFO_shape` column with the word `"circle"` in it, you get separate columns like `UFO_shape_circle`, `UFO_shape_triangle`, etc. — each filled with 0s and 1s.

We do this for `UFO_shape` and our new `city_class` column.

```python
df_encoded = pd.get_dummies(df, columns=['UFO_shape', 'city_class'], drop_first=False)

# pandas creates True/False by default — convert those to 1/0
bool_cols = df_encoded.select_dtypes(include='bool').columns
df_encoded[bool_cols] = df_encoded[bool_cols].astype(int)

print(df_encoded.shape)
df_encoded.head()
```

```python
# Save the fully encoded dataset (0s and 1s, ready for further analysis or ML)
df_encoded.to_csv('ufo_encoded_final.csv', index=False)
print('Saved: ufo_encoded_final.csv')
```

## Step 5: Final result — which location has which shape, and for how long

This is the actual answer to the question: *"which locations are more likely to have what shape of UFO and for how long?"* We group the data by `city_class` and, for each group, show:
- how many sightings happened there
- the most commonly reported shape
- the average (and median) duration of the encounter

```python
result = (
    df.groupby('city_class')
    .agg(
        num_sightings=('UFO_shape', 'count'),
        most_common_shape=('UFO_shape', lambda x: x.value_counts().idxmax()),
        avg_duration_seconds=('length_of_encounter_seconds', 'mean'),
        median_duration_seconds=('length_of_encounter_seconds', 'median'),
    )
    .sort_values('num_sightings', ascending=False)
    .reset_index()
)

result['avg_duration_seconds'] = result['avg_duration_seconds'].round(0)
result
```

**Note:** `avg_duration_seconds` (mean) can be skewed by a single extreme outlier in a small group — that's why `median_duration_seconds` is included too. If the two numbers are very different for a location, that's a sign of an outlier, not a reliable pattern.

Also note: `Other` and `Unknown` will have the most rows since most cities only appear once and get grouped together — the named individual cities (Los Angeles, Chicago, etc.) are the ones with enough data per group to actually trust the pattern.

```python
result.to_csv('ufo_location_shape_duration_summary.csv', index=False)
print('Saved: ufo_location_shape_duration_summary.csv')
```

## Step 6 (bonus): Relationship between location and shape

A cross-tab shows, for each state, how many sightings of each shape were reported. This is another way to see the relationship your supervisor asked about — which location tends to go with which shape.

```python
# Cross-tab: state vs shape (counts)
state_shape_crosstab = pd.crosstab(df['state'], df['UFO_shape'])
state_shape_crosstab
```
