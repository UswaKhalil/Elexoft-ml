```python
import csv
import numpy as np

filename = "../Day 4/eurepoc_global_dataset_.csv"

scores = []

with open(filename, "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        score = row["impact_indicator_score"]

        if score != "Not available" and score != "":
            scores.append(int(score))

arr = np.array(scores[:20])

print(arr)
```

```python
#insepction
print("Array:", arr)
print("Shape:", arr.shape)
print("Number of dimensions:", arr.ndim)
print("Total elements:",arr.size)
print("Data type:",arr.dtype)
```

```python
matrix = arr.reshape(4, 5)

print(matrix)
print("shape:", matrix.shape)
```

```python
print("one value")
print(matrix[1, 2])
print("entire row")
print(matrix[0, :])
print("entire column")
print(matrix[:, 0])
print("multiple rows")
print(matrix[1:3, :])
print("multiple columns")
print(matrix[:, 1:4])
print("smaller section")
print(matrix[1:3, 1:4])
```

```python
new_matrix =  matrix + 1
print("original:")
print(matrix)

print("after adding 1:")
print(new_matrix)
```
