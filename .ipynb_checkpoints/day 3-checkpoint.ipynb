 ## Cell 3: Code (Basic Math & Variables)

```python
# Define sample variables
items_sold = 150
unit_price = 25.50

# Calculate total revenue
total_revenue = items_sold * unit_price

# Format and display the output
print(f"Total Items Sold: {items_sold}")
print(f"Projected Revenue: ${total_revenue:,.2f}")
```

```python
# Create a dummy sales dataset
data = {
    "Product": ["Gadgets", "Widgets", "Gizmos", "Tools", "Devices"],
    "Sales_Q1": [120, 85, 150, 60, 110],
    "Sales_Q2": [140, 95, 130, 80, 125],
}

# Load into a Pandas DataFrame
df = pd.DataFrame(data)

# Calculate a new column
df["Total_Sales"] = df["Sales_Q1"] + df["Sales_Q2"]

# Simply type the variable name at the end of a cell to display it cleanly
df
```

```python
# Set up the visualization style
plt.style.use("seaborn-v0_8-whitegrid")
fig, ax = plt.subplots(figsize=(8, 5))

# Plot a bar chart
ax.bar(df["Product"], df["Total_Sales"], color="skyblue", edgecolor="black")

# Add design elements
ax.set_title("Total Sales Performance by Product", fontsize=14, fontweight="bold")
ax.set_xlabel("Product Category", fontsize=12)
ax.set_ylabel("Units Sold", fontsize=12)

# Display the plot directly inside the notebook
plt.show()
```

```python
def calculate_total(price,quantity, tax_rates=0.5):
    subtotal = price * quantity
    total = subtotal + (subtotal * tax_rates)
    return total
print (calculate_total(100, 2))
print(calcuate_total(100, 2, 0.10))
```

```python
def is_even(number):
    return number % 2 == 0
def square(number):
    return number * number 
def average (a, b, c):
    return(a + b + c)/ 3
def to_uppercase(text):
    return text.upper()
def count_vowles(text):
    vowels = "aeiouAEIOU"
    count = 0
    for letter in text:
        if letter in vowels:
            count = count + 1
    return count
    
print(is_even(8))
print(square(5))
print(average(2, 4, 9))
print(to_uppercase("machine learning"))
print(count_vowles("understanding python"))
```

```python
contacts = {}

def add_contact(name, phone):
    contacts[name] = phone
    print(name, "has been added to your contact book")

def search_contact(name):
    if name in contacts:
        print(name, "is at", contacts[name])
    else:
        print(name, "was not found in your contact book")

def delete_contact(name):
    if name in contacts:
        del contacts[name]
        print(name, "has been removed")
    else:
        print(name, "was not found, nothing to remove")

def show_all_contacts():
    for name, phone in contacts.items():
        print(name, ":", phone)

add_contact("aisha", "03001234567")
add_contact("bilal", "03007654321")
search_contact("aisha")
delete_contact("bilal")
show_all_contacts()

#added functions:
def update_contact(name, phone):
    if name in contacts:
         contacts[name] = phone
         print(name, "has been updated")
    else:
        print(name, "was not found in your contact book")

def contact_count():
    return len(contacts)

update_contact("aisha", "03111234567")
print("Total contacts:", contact_count())
```

```python
class Dog:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def bark(self):
    print(self.name + " says Woof!")

d1 = Dog("Buddy", 3)
d1.bark()
```
