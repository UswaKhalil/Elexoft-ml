```python
d2 = titanic.dropna(subset=['Age']).copy()

def categorize(row):
    if row['Age'] < 16:
        return 'Child'
    elif row['Sex'] == 'female':
        return 'Woman'
    else:
        return 'Man'

d2['Group'] = d2.apply(categorize, axis=1)
print(d2['Group'].value_counts())

table2 = d2.groupby(['Pclass','Group'])['Survived'].agg(['sum','count','mean'])
table2.columns = ['survivors','total','survival_rate']
table2
```

```python
pivot3 = d2.groupby(['Pclass','Group'])['Survived'].mean().unstack()[['Woman','Child','Man']]
pivot3['Woman_Man_gap_pts'] = (pivot3['Woman'] - pivot3['Man']) * 100
pivot3
```

```python
#Woman + Child = priority g, compared with Man, by class
priority = d2[d2['Group'].isin(['Woman','Child'])]
men = d2[d2['Group'] == 'Man']

summary5 = pd.DataFrame({
    'priority_survival': priority.groupby('Pclass')['Survived'].mean(),
    'priority_n': priority.groupby('Pclass')['Survived'].count(),
    'men_survival': men.groupby('Pclass')['Survived'].mean(),
    'men_n': men.groupby('Pclass')['Survived'].count(),
})
summary5
```

```python
import pandas as pd

titanic = pd.read_csv("Titanic-Dataset1.csv")

#solo vs grouped, restrict to 3rd class
ticket_counts = titanic['Ticket'].value_counts()
titanic['TicketGroupSize'] = titanic['Ticket'].map(ticket_counts)
c3 = titanic[titanic['Pclass'] == 3].copy()
c3['Solo'] = c3['TicketGroupSize'] == 1
```

```python
#solo women survived
women_solo_survived = c3[
    (c3['Sex'] == 'female') & 
    (c3['Solo'] == True) & 
    (c3['Survived'] == 1)]

print("WOMEN, SOLO, SURVIVED:", len(women_solo_survived))
women_solo_survived[['Name','Age','Survived']]
```

```python
#grouped women survived
women_grouped_survived =  c3[
    (c3['Sex'] == 'female') & 
    (c3['Solo'] == False) & 
    (c3['Survived'] == 1)]
print("WOMEN, GROUPED, SURVIVED:", len(women_grouped_survived))
women_grouped_survived[['Name','Age','Survived']]
```

```python
#solo men survuved
men_solo_survived = c3[
    (c3['Sex'] == 'male') & 
    (c3['Solo'] == True) & 
    (c3['Survived'] == 1)]
print("MEN, SOLO, SURVIVED:", len(men_solo_survived))
men_solo_survived[['Name','Age','Survived']]
```

```python
#grouped men survived
men_grouped_survived = c3[
    (c3['Sex'] == 'male') & 
    (c3['Solo'] == False) & 
    (c3['Survived'] == 1)]
print("MEN, GROUPED, SURVIVED:", len(men_grouped_survived))
men_grouped_survived[['Name','Age','Survived']]
```

```python
# --- Build the comparison table ---
comparison = pd.DataFrame({
    'Group': ['Women - Solo', 'Women - Grouped', 'Men - Solo', 'Men - Grouped'],
    'Survivors': [
        len(women_solo_survived),
        len(women_grouped_survived),
        len(men_solo_survived),
        len(men_grouped_survived)
    ],
    'Total': [
        len(c3[(c3['Sex']=='female') & (c3['Solo']==True)]),
        len(c3[(c3['Sex']=='female') & (c3['Solo']==False)]),
        len(c3[(c3['Sex']=='male') & (c3['Solo']==True)]),
        len(c3[(c3['Sex']=='male') & (c3['Solo']==False)])
    ]
})

comparison['Survival Rate %'] = (comparison['Survivors'] / comparison['Total'] * 100).round(1)
comparison
```
