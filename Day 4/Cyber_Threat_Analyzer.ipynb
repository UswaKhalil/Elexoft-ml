```python
def show_total_incidents():
    total_incidents = 0

    with open(filename, "r", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        for row in reader:
            total_incidents += 1

    print("Total incidents:", total_incidents)
```

```python
incidents_by_year = {}

with open(filename, "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        date = row["start_date"]

        if date != "Not available" and "." in date:
            year = date.split(".")[-1]

            if year.isdigit():
                if year in incidents_by_year:
                    incidents_by_year[year] += 1
                else:
                    incidents_by_year[year] = 1

print(incidents_by_year)
```

```python
def disruption_count():
    disruption_total = 0

    with open(filename, "r", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        for row in reader:
            incident_type = row["incident_type"]

            if "Disruption" in incident_type:
                disruption_total += 1

    print("Incidents involving disruption:", disruption_total)
```

```python
def data_theft_count():
    data_theft_total = 0

    with open(filename, "r", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        for row in reader:
            data_theft = row["data_theft"]

            if "information" in data_theft:
                data_theft_total += 1

    print("Incidents involving data theft:", data_theft_total)
```

```python
categories = {}

with open(filename, "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        category = row["receiver_category"]

        if category != "Not available" and category != "":
            if category in categories:
                categories[category] += 1
            else:
                categories[category] = 1

top_categories = sorted(
    categories.items(),
    key=lambda item: item[1],
    reverse=True
)
```

```python
def search_incident(incident_id):
    try:
        with open(filename, "r", encoding="utf-8") as file:
            reader = csv.DictReader(file)

            for row in reader:
                if row["incident_id"] == incident_id:
                    print("\nIncident found!")
                    print("Name:", row["name"])
                    print("Country:", row["receiver_country"])
                    print("Type:", row["incident_type"])
                    return

            print("\nIncident not found.")

    except FileNotFoundError:
        print("Dataset file could not be found.")
```

```python
def save_report():
    with open("cyber_threat_report.txt", "w", encoding="utf-8") as file:
        file.write("CYBER INCIDENT ANALYSIS REPORT\n")
        file.write("==============================\n\n")
        file.write("Total incidents: 3416\n")
        file.write("Incidents involving disruption: 1475\n")
        file.write("Incidents involving data theft: 1669\n\n")

        file.write("Top 10 targeted countries:\n")

        for country, count in top_countries[:10]:
            file.write(country + ": " + str(count) + "\n")

        file.write("\nTop 10 target categories:\n")

        for category, count in top_categories[:10]:
            file.write(category + ": " + str(count) + "\n")

    print("Report saved successfully!")
```
