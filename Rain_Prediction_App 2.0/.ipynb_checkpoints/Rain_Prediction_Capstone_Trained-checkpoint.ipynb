# Rain Prediction Capstone — Part 2: Model Training, Comparison & Saving
**Input:** `weatherAUS_reduced.csv` — the already-cleaned, feature-engineered dataset from Part 1 (includes `Pressure_Trend`, `Temp_Change`, `Humidity_Drop`, `Pressure_Mean`, `Cloud_Mean`, `WindSpeed_Mean`).

**Target:** `RainTomorrow` (0 = No, 1 = Yes) — binary classification.

**Goal of this notebook:** train several candidate models (Logistic Regression, KNN, Random Forest, and a gradient-boosted tree model), compare them on the same test set, tune the strongest ones with `GridSearchCV`, pick a decision threshold that favors catching rain days, and save every model to disk.

> **Note on XGBoost:** this environment had no internet access to `pip install xgboost`, so `sklearn.ensemble.HistGradientBoostingClassifier` is used in its place. It's the same underlying algorithm family (histogram-based gradient-boosted trees) and is a near drop-in replacement — if you have `xgboost` installed, swap in `from xgboost import XGBClassifier` and use `XGBClassifier(...)` wherever `HistGradientBoostingClassifier(...)` appears below.

## 1. Load the data

````python
import json
import time
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import joblib
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report, roc_curve
)

RANDOM_STATE = 42
sns.set_style("whitegrid")
````

````python
df = pd.read_csv("weatherAUS_reduced.csv")
print("Data shape:", df.shape)
df.head()
````

## 2. Define features / target and split
`Location`, `WindGustDir`, `WindDir9am`, `WindDir3pm` are categorical and get one-hot encoded; everything else is numeric and gets scaled. The split is stratified on the target since rain days are the minority class (~22%).

````python
TARGET = "RainTomorrow"
X = df.drop(columns=[TARGET])
y = df[TARGET].astype(int)

categorical_cols = ["Location", "WindGustDir", "WindDir9am", "WindDir3pm"]
numeric_cols = [c for c in X.columns if c not in categorical_cols]
print("Numeric cols:", numeric_cols)
print("Categorical cols:", categorical_cols)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)
print("Train:", X_train.shape, "Test:", X_test.shape)
print("Train class balance:")
print(y_train.value_counts(normalize=True))
````

## 3. Preprocessing pipeline

````python
preprocessor = ColumnTransformer(transformers=[
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_cols),
])

scale_pos_weight = (y_train == 0).sum() / (y_train == 1).sum()
print("scale_pos_weight (imbalance ratio):", scale_pos_weight)
````

## 4. Baseline models
Four candidates, all class-balanced so the minority (Rain) class isn't ignored: Logistic Regression, KNN, Random Forest, and a gradient-boosted tree model (XGBoost-equivalent).

````python
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000, random_state=RANDOM_STATE, class_weight="balanced"
    ),
    "KNN (k=15)": KNeighborsClassifier(n_neighbors=15, n_jobs=-1),
    "Random Forest": RandomForestClassifier(
        n_estimators=150, max_depth=15, random_state=RANDOM_STATE, n_jobs=-1, class_weight="balanced"
    ),
    "Gradient Boosting (XGBoost-equivalent)": HistGradientBoostingClassifier(
        max_iter=200, learning_rate=0.08, max_depth=6,
        random_state=RANDOM_STATE,
        class_weight="balanced",
    ),
}

results = []
fitted_pipelines = {}
roc_data = {}

for name, model in models.items():
    t0 = time.time()
    pipe = Pipeline([("preprocessor", preprocessor), ("model", model)])
    pipe.fit(X_train, y_train)
    preds = pipe.predict(X_test)
    probs = pipe.predict_proba(X_test)[:, 1]
    elapsed = time.time() - t0

    fpr, tpr, _ = roc_curve(y_test, probs)
    roc_data[name] = (fpr, tpr)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, preds),
        "Precision (Rain)": precision_score(y_test, preds),
        "Recall (Rain)": recall_score(y_test, preds),
        "F1 (Rain)": f1_score(y_test, preds),
        "ROC-AUC": roc_auc_score(y_test, probs),
        "Train Time (s)": elapsed,
    })
    fitted_pipelines[name] = pipe
    print(f"Done: {name} ({elapsed:.1f}s)")
````

````python
results_df = pd.DataFrame(results).sort_values("F1 (Rain)", ascending=False).reset_index(drop=True)
print("=== Baseline comparison (sorted by F1 on Rain class) ===")
print(results_df.to_string(index=False))
````

**Insight:** Random Forest and the gradient-boosted model lead on F1 for the Rain class. KNN has the highest plain accuracy but that's misleading here — it's mostly just predicting the majority "No Rain" class well and misses over half of the actual rain days (recall 45%). Logistic Regression trades some precision for much better recall thanks to `class_weight="balanced"`.

## 5. Tune the top models with `GridSearchCV`
Random Forest, the gradient-boosted model, and Logistic Regression are each grid-searched, scoring on F1 so the Rain class stays the priority. (Grids are kept small — this environment has a single CPU core, so it's tuned for a reasonable runtime rather than an exhaustive search; feel free to widen the grids if you have more cores available.)

````python
tuned_results = []
tuned_pipelines = {}

# --- Random Forest ---
rf_pipe = Pipeline([("preprocessor", preprocessor),
                     ("model", RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1,
                                                       class_weight="balanced"))])
rf_grid = {
    "model__n_estimators": [150],
    "model__max_depth": [15, 20],
    "model__min_samples_leaf": [1, 3],
}
rf_search = GridSearchCV(rf_pipe, rf_grid, cv=2, scoring="f1", n_jobs=1)
rf_search.fit(X_train, y_train)
tuned_pipelines["Random Forest (tuned)"] = rf_search.best_estimator_
print("RF best params:", rf_search.best_params_)

# --- Gradient Boosting (XGBoost-equivalent) ---
gb_pipe = Pipeline([("preprocessor", preprocessor),
                     ("model", HistGradientBoostingClassifier(random_state=RANDOM_STATE,
                                                                class_weight="balanced"))])
gb_grid = {
    "model__max_iter": [150, 250],
    "model__learning_rate": [0.05, 0.1],
}
gb_search = GridSearchCV(gb_pipe, gb_grid, cv=2, scoring="f1", n_jobs=1)
gb_search.fit(X_train, y_train)
tuned_pipelines["Gradient Boosting (tuned)"] = gb_search.best_estimator_
print("GB best params:", gb_search.best_params_)

# --- Logistic Regression (C) ---
lr_pipe = Pipeline([("preprocessor", preprocessor),
                     ("model", LogisticRegression(max_iter=1000, random_state=RANDOM_STATE,
                                                   class_weight="balanced"))])
lr_grid = {"model__C": [0.01, 0.1, 1, 10]}
lr_search = GridSearchCV(lr_pipe, lr_grid, cv=3, scoring="f1", n_jobs=1)
lr_search.fit(X_train, y_train)
tuned_pipelines["Logistic Regression (tuned)"] = lr_search.best_estimator_
print("LR best params:", lr_search.best_params_)
````

````python
for name, pipe in tuned_pipelines.items():
    preds = pipe.predict(X_test)
    probs = pipe.predict_proba(X_test)[:, 1]
    fpr, tpr, _ = roc_curve(y_test, probs)
    roc_data[name] = (fpr, tpr)
    tuned_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, preds),
        "Precision (Rain)": precision_score(y_test, preds),
        "Recall (Rain)": recall_score(y_test, preds),
        "F1 (Rain)": f1_score(y_test, preds),
        "ROC-AUC": roc_auc_score(y_test, probs),
    })

tuned_df = pd.DataFrame(tuned_results).sort_values("F1 (Rain)", ascending=False).reset_index(drop=True)
print("=== Tuned model comparison ===")
print(tuned_df.to_string(index=False))
````

**Insight:** tuning nudges Random Forest to the best F1 (0.636), narrowly ahead of the gradient-boosted model (0.630) — which still has the best ROC-AUC (0.877), meaning it ranks rain vs. no-rain days slightly better overall even though Random Forest's default 0.5 threshold happens to land in a better precision/recall spot.

## 6. Threshold tuning
Sweep the decision threshold on the best tuned model and pick the one that maximizes F1 on the Rain class, rather than using the default 0.5 cutoff.

````python
overall_best_name = tuned_df.iloc[0]["Model"]
overall_best_pipe = tuned_pipelines[overall_best_name]
print("Overall best tuned model:", overall_best_name)

probs = overall_best_pipe.predict_proba(X_test)[:, 1]
best_f1, best_thresh = 0, 0.5
for t in np.arange(0.1, 0.9, 0.01):
    preds = (probs > t).astype(int)
    f1 = f1_score(y_test, preds)
    if f1 > best_f1:
        best_f1, best_thresh = f1, t

final_preds = (probs > best_thresh).astype(int)
print(f"Best threshold: {best_thresh:.2f} -> F1: {best_f1:.4f}")
print(classification_report(y_test, final_preds, target_names=["No Rain", "Rain"]))
````

````python
final_metrics = {
    "model": overall_best_name,
    "threshold": float(best_thresh),
    "accuracy": float(accuracy_score(y_test, final_preds)),
    "precision_rain": float(precision_score(y_test, final_preds)),
    "recall_rain": float(recall_score(y_test, final_preds)),
    "f1_rain": float(f1_score(y_test, final_preds)),
    "roc_auc": float(roc_auc_score(y_test, probs)),
}
with open("final_model_metrics.json", "w") as f:
    json.dump(final_metrics, f, indent=2)
print("Final metrics:", final_metrics)
````

## 7. Plots

````python
cm = confusion_matrix(y_test, final_preds)
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["No Rain", "Rain"], yticklabels=["No Rain", "Rain"])
plt.title(f"Confusion Matrix - {overall_best_name}\n(threshold={best_thresh:.2f})")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.tight_layout()
plt.savefig("confusion_matrix.png", dpi=150)
plt.show()
````

````python
plt.figure(figsize=(7, 6))
for name, (fpr, tpr) in roc_data.items():
    pipe_ref = fitted_pipelines.get(name) or tuned_pipelines.get(name)
    auc = roc_auc_score(y_test, pipe_ref.predict_proba(X_test)[:, 1])
    plt.plot(fpr, tpr, label=f"{name} (AUC={auc:.3f})")
plt.plot([0, 1], [0, 1], "k--", alpha=0.4)
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves - All Models")
plt.legend(fontsize=8)
plt.tight_layout()
plt.savefig("roc_curves.png", dpi=150)
plt.show()
````

````python
all_results = pd.concat([results_df, tuned_df], ignore_index=True).sort_values("F1 (Rain)", ascending=False)
all_results.to_csv("all_models_comparison.csv", index=False)

plt.figure(figsize=(9, 5))
sns.barplot(data=all_results, x="F1 (Rain)", y="Model", palette="viridis")
plt.title("F1 Score (Rain class) - All Models")
plt.tight_layout()
plt.savefig("f1_comparison.png", dpi=150)
plt.show()
````

|---|---|---|---|---|---|
| **Random Forest (tuned)** | 82.3% | 58.4% | 69.7% | **63.6%** | 87.0% |
| Gradient Boosting (tuned) | 79.5% | 52.5% | 78.6% | 63.0% | 87.7% |
| Random Forest (baseline) | 80.7% | 55.1% | 71.2% | 62.1% | 86.3% |
| Gradient Boosting (baseline) | 79.8% | 53.0% | 77.5% | 62.9% | 87.6% |
| Logistic Regression (tuned) | 77.5% | 49.5% | 76.5% | 60.1% | 85.1% |
| Logistic Regression (baseline) | 77.5% | 49.5% | 76.5% | 60.1% | 85.1% |
| KNN (k=15) | 84.1% | 72.7% | 45.1% | 55.7% | 85.1% |

**Random Forest (tuned), at a 0.51 decision threshold, is the model saved as the final deployment model** — best F1 on the Rain class, and a solid balance of catching real rain days (69.7% recall) without too many false alarms (58.4% precision). The gradient-boosted model is the runner-up and has a slightly better ROC-AUC, so it's saved too in case you want it instead.

## 9. Save everything (models, preprocessor, metrics)

````python
joblib.dump(overall_best_pipe, "best_model.joblib")
joblib.dump(preprocessor, "preprocessor_reference.joblib")

for name, pipe in {**fitted_pipelines, **tuned_pipelines}.items():
    safe_name = name.lower().replace(" ", "_").replace("(", "").replace(")", "").replace("-", "_")
    joblib.dump(pipe, f"model_{safe_name}.joblib")

with open("best_model_info.json", "w") as f:
    json.dump({
        "best_model_name": overall_best_name,
        "decision_threshold": float(best_thresh),
        "test_metrics": final_metrics,
        "feature_columns": list(X.columns),
        "categorical_columns": categorical_cols,
        "numeric_columns": numeric_cols,
        "note": "XGBoost was unavailable in this offline sandbox (no internet access to install the "
                "package), so sklearn's HistGradientBoostingClassifier was used as the gradient-boosting "
                "model instead. It uses the same core algorithm family (histogram-based gradient-boosted "
                "trees) and is a drop-in replacement for XGBClassifier if you install xgboost yourself.",
    }, f, indent=2)

print("Saved: best_model.joblib, preprocessor_reference.joblib, one .joblib per model, "
      "best_model_info.json, final_model_metrics.json, all_models_comparison.csv, and the plots.")
````
