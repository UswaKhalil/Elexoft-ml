# Titanic Dataset — Deep-Dive Analysis

Goes beyond the obvious "1st class survived more than 3rd class" and looks for group-level, non-obvious survival patterns: ticket-sharing groups, family clustering, and interaction effects between class, age, sex, and traveling alone vs. grouped.

**Structure:**
1. Setup & data check
2. Baseline (the obvious stuff, so we know what's *not* new)
3. Finding 1 — Ticket-group survival effect (odds ratio, logistic regression, cross-validation)
4. Finding 2 — Family/ticket-group clustering test (chi-square vs. independence)
5. Finding 3 — Children vs adults, by class
6. Finding 4 — Solo vs grouped survival, by sex, 3rd class only


## 1. Setup & data check

```python
import pandas as pd
import numpy as np
from scipy.stats import chisquare, chi2_contingency
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score

pd.set_option('display.width', 120)

df = pd.read_csv('Titanic-Dataset.csv')
print(df.shape)
df.head()
```

```python
# Always check missing data before trusting any column
df.isnull().sum()
```

`Age` is missing for 177 of 891 rows (20%). `Cabin` is missing for 687 (77% — too sparse to use directly). `Embarked` is missing for 2. Keep this in mind: any analysis using `Age` will run on a smaller, filtered subset (714 rows).

## 2. Baseline — the obvious result first

Establishing this first means any later finding has to explain something *beyond* this.

```python
df.groupby(['Pclass','Sex'])['Survived'].agg(['mean','count'])
```

## 3. Finding 1 — Does your ticket-group's fate predict your own survival?

**Idea:** People sharing a `Ticket` number usually traveled/boarded together. For each passenger, compute the survival rate of *everyone else* on their ticket (excluding themselves, to avoid circularity).

```python
ticket_counts = df['Ticket'].value_counts()
df['TicketGroupSize'] = df['Ticket'].map(ticket_counts)

def group_other_survival(row, df):
    others = df[(df['Ticket'] == row['Ticket']) & (df['PassengerId'] != row['PassengerId'])]
    if len(others) == 0:
        return np.nan
    return others['Survived'].mean()

df['OthersSurvivalRate'] = df.apply(lambda r: group_other_survival(r, df), axis=1)
df[['PassengerId','Ticket','TicketGroupSize','Survived','OthersSurvivalRate']].head(10)
```

```python
# Bucket into: solo / grouped-but-others-died / grouped-and-others-lived
solo = df[df['TicketGroupSize'] == 1]
grouped = df[df['TicketGroupSize'] > 1].copy()
grouped['OthersSurvived'] = grouped['OthersSurvivalRate'] > 0.5

print('Solo travelers:            n=%d  survival=%.1f%%' % (len(solo), solo['Survived'].mean()*100))
print('Grouped, others mostly died: n=%d  survival=%.1f%%' % (
    len(grouped[~grouped['OthersSurvived']]), grouped[~grouped['OthersSurvived']]['Survived'].mean()*100))
print('Grouped, others mostly lived: n=%d  survival=%.1f%%' % (
    len(grouped[grouped['OthersSurvived']]), grouped[grouped['OthersSurvived']]['Survived'].mean()*100))
```

### Stress test: is this just class/sex/age/family-size in disguise?

Run a logistic regression with `OthersSurvived` alongside the standard demographic predictors. If its coefficient survives, it's adding real information, not just re-stating who's rich/female/young.

```python
df['SexNum'] = (df['Sex'] == 'female').astype(int)
df['AgeFilled'] = df['Age'].fillna(df['Age'].median())
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

sub = df.dropna(subset=['OthersSurvivalRate']).copy()
sub['OthersSurvived'] = (sub['OthersSurvivalRate'] > 0.5).astype(int)

X = sub[['OthersSurvived','SexNum','Pclass','AgeFilled','FamilySize']]
y = sub['Survived']

model = LogisticRegression(max_iter=1000).fit(X, y)
coefs = pd.Series(model.coef_[0], index=X.columns)
odds_ratios = np.exp(coefs)
pd.DataFrame({'coef (log-odds)': coefs, 'odds ratio': odds_ratios})
```

```python
# Chi-square significance test on the raw 2x2 table
ct = pd.crosstab(sub['OthersSurvived'], sub['Survived'])
print(ct)
chi2, p, dof, exp = chi2_contingency(ct)
print('p-value:', p)
```

```python
# Does it actually improve prediction (not just correlate)?
X_without = sub[['SexNum','Pclass','AgeFilled','FamilySize']]
score_with = cross_val_score(LogisticRegression(max_iter=1000), X, y, cv=5).mean()
score_without = cross_val_score(LogisticRegression(max_iter=1000), X_without, y, cv=5).mean()
print('CV accuracy WITH OthersSurvived:', round(score_with, 4))
print('CV accuracy WITHOUT:            ', round(score_without, 4))
```

**Takeaway:** Odds ratio ≈ 6.4, p ≈ 1e-11, and it adds ~3.5 points of cross-validated accuracy on top of sex+class+age+family size. Survival was heavily a *group* event, not purely individual.

## 4. Finding 2 — Do families/ticket-groups sink or swim together, more than chance predicts?

**Idea:** For every ticket-group with 2+ people, check whether the outcome was uniform (all survived, or all died) vs. mixed. Compare that to what you'd expect if every person's survival were an independent coin flip at the group's own average rate.

```python
fam = df[df['FamilySize'] > 1].copy()
p_overall = fam['Survived'].mean()

fam_groups = fam.groupby('Ticket')['Survived'].agg(['mean','count'])
fam_groups = fam_groups[fam_groups['count'] > 1]

observed = [
    (fam_groups['mean'] == 1.0).sum(),                                  # all survived
    (fam_groups['mean'] == 0.0).sum(),                                  # all died
    ((fam_groups['mean'] > 0) & (fam_groups['mean'] < 1)).sum(),        # mixed
]
total_groups = len(fam_groups)
print('Observed [all-survived, all-died, mixed]:', observed, ' total groups:', total_groups)
```

```python
# Expected counts under independence, using each group's own size
exp_all_survive = exp_all_die = exp_mixed = 0
for s in fam_groups['count']:
    exp_all_survive += p_overall**s
    exp_all_die     += (1 - p_overall)**s
    exp_mixed       += 1 - p_overall**s - (1 - p_overall)**s

expected = [exp_all_survive, exp_all_die, exp_mixed]
print('Expected under independence:', [round(x,1) for x in expected])

chi2, p = chisquare(observed, f_exp=expected)
print('chi2 =', round(chi2,2), ' p-value =', p)

uniform_observed = (observed[0] + observed[1]) / total_groups
uniform_expected = (expected[0] + expected[1]) / total_groups
print('Uniform-outcome rate, observed:', round(uniform_observed,3))
print('Uniform-outcome rate, expected by chance:', round(uniform_expected,3))
```

**Takeaway:** χ² ≈ 19.8, p ≈ 0.00005 — groups were far more "all-or-nothing" (61% uniform) than pure chance would produce (41% uniform). Survival clustered by group, not by independent individual luck.

## 5. Finding 3 — Children vs. adults, by class

**Idea:** Does being a child protect you equally in every class, or does the 'women and children first' effect weaken in steerage?

```python
# Age has missing values - drop them rather than impute, so we don't fake who counts as a child
d = df.dropna(subset=['Age']).copy()
print('Usable rows:', len(d), 'out of', len(df))

d['IsChild'] = d['Age'] < 16
table = d.groupby(['Pclass','IsChild'])['Survived'].agg(['sum','count','mean'])
table.columns = ['survivors','total','survival_rate']
table
```

```python
# Reshape to compare children vs adults side by side, per class
pivot = d.groupby(['Pclass','IsChild'])['Survived'].mean().unstack()
pivot.columns = ['Adult','Child']
pivot['child_advantage_pts'] = (pivot['Child'] - pivot['Adult']) * 100
pivot
```

**Takeaway:** every class shows a child survival bonus, but it's ~2.5x bigger in 2nd class (+58 pts) than in 3rd class (+23 pts). Note 1st class only has 6 children total — treat that number cautiously.

## 6. Finding 4 — Solo vs. grouped survival, by sex, 3rd class only

**Idea:** Restrict to 3rd class and check whether traveling alone (by ticket) helps or hurts, separately for men and women.

```python
c3 = df[df['Pclass'] == 3].copy()
c3['Solo'] = c3['TicketGroupSize'] == 1

result = c3.groupby(['Sex','Solo'])['Survived'].agg(['sum','count','mean'])
result.columns = ['survivors','total','survival_rate']
result
```

```python
pivot2 = c3.groupby(['Sex','Solo'])['Survived'].mean().unstack()
pivot2.columns = ['Grouped','Solo']
pivot2['solo_effect_pts'] = (pivot2['Solo'] - pivot2['Grouped']) * 100
pivot2
```

**Takeaway:** in 3rd class, solo women survived *less* than grouped women (companionship helped), while solo men survived *more* than grouped men (the opposite pattern) — a genuine sex-by-grouping interaction, not just noise.
