```python
import pandas as pd

df = pd.read_csv('cybersecurity_intrusion_data.csv')
print(df.shape)
df.head()
```

```python
# MISSING VALUES
```

```python
df.isna().sum()
```

```python
# only encryption_used has missing values
# treating missing as "no encryption was used" instead of dropping rows

df['encryption_used'] = df['encryption_used'].fillna('None')

print('Missing encryption_used values now:', df['encryption_used'].isna().sum())
df['encryption_used'].value_counts()
```

```python
# DUPLICATES
```

```python
dupes = df.drop(columns=['session_id']).duplicated().sum()
print('Duplicate rows (excluding session_id):', dupes)
```

```python
# drop session_id, its just an ID for each row, not useful for finding patterns

df = df.drop(columns=['session_id'])
df.head()
```

```python
## turn the text columns into 0s and 1s (one-hot encoding)
```

```python
df_encoded = pd.get_dummies(df, columns=['protocol_type', 'encryption_used', 'browser_type'], drop_first=False)

# True/False by default, convert those to 1/0
bool_cols = df_encoded.select_dtypes(include='bool').columns
df_encoded[bool_cols] = df_encoded[bool_cols].astype(int)

print(df_encoded.shape)
df_encoded.head()
```

```python
df_encoded.to_csv('cybersecurity_encoded_final.csv', index=False)
print('Saved: cybersecurity_encoded_final.csv')
```

```python
# RESULT
```

```python
# compare average values for attacked sessions (1) vs normal sessions (0)

result = (
    df.groupby('attack_detected')
    .agg(
        num_sessions=('failed_logins', 'count'),
        avg_failed_logins=('failed_logins', 'mean'),
        avg_login_attempts=('login_attempts', 'mean'),
        avg_session_duration=('session_duration', 'mean'),
        avg_ip_reputation_score=('ip_reputation_score', 'mean'),
        avg_packet_size=('network_packet_size', 'mean'),
        pct_unusual_time_access=('unusual_time_access', 'mean'),
    )
    .round(2)
    .reset_index()
)

result
```

Just from this table you can already see a pattern: attacked sessions (`attack_detected` = 1) have way more `failed_logins` and `login_attempts` on average, a lower `ip_reputation_score`, and access unusual times a lot more often than normal sessions.

```python
# CORRELATION WITH attack_detected
```

```python
# how strongly does each column move together with attack_detected?
# closer to 1 or -1 = stronger relationship, closer to 0 = weak/no relationship

corr = df_encoded.corr()['attack_detected'].drop('attack_detected').sort_values(key=abs, ascending=False)
corr
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 7))
colors = corr.apply(lambda v: 'crimson' if v > 0 else 'steelblue')
plt.barh(corr.index, corr.values, color=colors)
plt.xlabel('Correlation with attack_detected')
plt.title('Which characteristics correlate most with an attack?')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

The columns at the top (`failed_logins`, `login_attempts`, `session_duration`, `ip_reputation_score`, `encryption_used_None`, `unusual_time_access`) have the strongest correlation with `attack_detected`. Everything near the bottom (packet size, protocol type, browser type) barely moves with attacks at all — so they're weak predictors.

```python
# ATTACK RATE BY CATEGORY (protocol, encryption, browser)
```

```python
# instead of just counts, look at what % of sessions in each category were attacks
# this shows if a category itself is risky, regardless of how common it is

for col in ['protocol_type', 'encryption_used', 'browser_type']:
    print(f'--- Attack rate by {col} ---')
    print(df.groupby(col)['attack_detected'].mean().round(3).sort_values(ascending=False))
    print()
```

No encryption (`encryption_used = None`) clearly has the highest attack rate compared to AES or DES — this matches what the correlation showed. Protocol type and browser type don't show much difference between categories, so they're not strong predictors on their own.

```python
# VISUAL: strongest predictor (failed_logins) vs attack_detected
```

```python
import seaborn as sns

plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x='attack_detected', y='failed_logins')
plt.xticks([0, 1], ['No Attack', 'Attack'])
plt.title('Failed Logins: Attack vs No Attack Sessions')
plt.xlabel('')
plt.ylabel('Failed Logins')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x='attack_detected', y='ip_reputation_score')
plt.xticks([0, 1], ['No Attack', 'Attack'])
plt.title('IP Reputation Score: Attack vs No Attack Sessions')
plt.xlabel('')
plt.ylabel('IP Reputation Score')
plt.tight_layout()
plt.show()
```

```python
result.to_csv('cybersecurity_result.csv', index=False)
print('Saved!')
```

## Answer

The strongest predictors of a cybersecurity attack in this dataset are:

1. **`failed_logins`** — attacked sessions have noticeably more failed login attempts
2. **`login_attempts`** — goes hand in hand with failed logins, both point to brute-force type behavior
3. **`ip_reputation_score`** — attacked sessions tend to come from lower-reputation IPs
4. **`session_duration`** — attacked sessions differ a lot in typical duration
5. **`encryption_used` (no encryption)** — sessions with no encryption are attacked more often
6. **`unusual_time_access`** — accessing at unusual times is more common in attacked sessions

`network_packet_size`, `protocol_type`, and `browser_type` showed weak or almost no relationship with `attack_detected`, so they're not strong predictors on their own.
